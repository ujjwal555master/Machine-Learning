```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
y.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

```python
x_train.shape, y_train.shape, y_test.shape, x_test.shape
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
y_predict = lr.predict(x_test)
r2_score(y_test, y_predict)
```

```python
import random
class MBGDregressor:
    def __init__(self, epoc, batch_size, lrt):
        self.epoc = epoc
        self.lrt = lrt
        self.intercept_ = None
        self.coeff_ = None
        self.batch_size = batch_size

    def fit(self, x_train, y_train):
        self.intercept_ = 0
        self.coeff_ = np.ones(x_train.shape[1])

        #calculate the intercept or beta note
        for i in range(self.epoc):

            for j in range(int(x_train.shape[0]/self.batch_size)):

                idx = random.sample(range(x_train.shape[0]),self.batch_size)
                
                y_hat = self.intercept_ + np.dot(x_train[idx], self.coeff_)
                lr_der = -2 * np.mean(y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lrt * lr_der)
            
                coef_dr = (-2 * np.dot((y_train[idx] - y_hat),x_train[idx])) / x_train.shape[0]
                self.coeff_ = self.coeff_ - (self.lrt * coef_dr)
        print(self.intercept_, self.coeff_)

    def predict(self, x_test):
        return np.dot(x_test,self.coeff_) + self.intercept_
```

```python
gd = MBGDregressor(epoc=300, batch_size=int(x_train.shape[0]/20), lrt=0.7)
```

```python
gd.fit(x_train, y_train)
```

```python
y_pred = gd.predict(x_test)
```

```python
r2_score(y_test, y_pred)
```
