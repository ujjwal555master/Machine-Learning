```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score 
```

```python
from sklearn.datasets import load_diabetes
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
X.shape
y.shape
```

```python
lr1 = LinearRegression()
svr = SVR()
rt1 = RandomForestRegressor(max_samples=0.75)
```

```python
estimators = [('lr',lr1), ('svr', svr), ('rt1', rt1)]
```

```python
for estimator in estimators:
    score = cross_val_score(estimator[1], X, y, scoring='r2', cv=10)
    print(estimator[0], np.round(np.mean(score),2))
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
vr = VotingRegressor(estimators)
```

```python
score = cross_val_score(vr, X, y, cv=10, scoring='r2')
np.round(np.mean(score), 2)
```

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vc = VotingRegressor(estimators=estimators,weights = [i,j,k])
            x = cross_val_score(vc,X,y,cv=10,scoring ='r2')
            print("for i={},j={},k={}".format(i,j,k),np.round(np.mean(x),2))
```
