```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.model_selection import GridSearchCV
from sklearn.datasets import load_diabetes
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
dt1 = DecisionTreeRegressor(criterion='friedman_mse', max_depth=5,max_features= 0.75,min_samples_split= 0.25)
```

```python
X_train.shape
```

```python
dt1.fit(X_train, y_train)
```

```python
y_pred = dt1.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```

# Hyper parameter turning

```python
pram = {
    'max_depth':[1, 5, 10, 15, 20, 25, 30, None],
    'criterion':['friedman_mse', 'poisson', 'squared_error', 'absolute_error'],
    'max_features':[0.25, 0.5, 0.75, 1.0],
    'min_samples_split':[0.25, 0.5, 0.75, 1.0]
}
```

```python
reg = GridSearchCV(DecisionTreeRegressor(), param_grid=pram)
```

```python
reg.fit(X_train, y_train)
```

```python
reg.best_score_
```

```python
reg.best_params_
```
